```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
data = pd.read_csv("Teen_Mental_Health_Dataset.csv")

data.head()
```

```python
print(max(data["age"]), min(data["age"]))
```

```python
males = data[data["gender"] == "male"]
females = data[data["gender"] == "female"]

bins_list = np.arange(13, 21)

fig, ax = plt.subplots(nrows=1, ncols=2, figsize=(15, 6))

ax[0].hist(males["age"], bins=bins_list, histtype="step", align="left")
ax[1].hist(females["age"], bins=bins_list, histtype="step", align="left")
```
