```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.cm as cm
import os

# number of data
N=5000
# size of each sample
L=4
# box size
B=100

!mkdir DATA
```

```python
# Function used to calculate the labels
def f(x):
    Q=np.heaviside(np.sin(np.pi*(x[1]+2*x[0])*(2*x[1]-x[0])/2000),0)
    return Q

def filename(s):
    return "DATA/"+s+".dat"

# random seed for reproducibility (I set the seed)
np.random.seed(12345)
print(np.random.randint(0,10),np.random.randint(0,10),np.random.randint(0,10))


# Generates a matrix x of uniformly sampled values:
# samples in the interval [0, 1], sums an offset of -0.5, then multiply it by B
# N is the number of data points, L is the number of features
# Generates the vector of labels initialized in 0
x,y = (np.random.random((N,L))-0.5)*B, np.zeros(N).astype(int)
for i in range(N):
    # label data according to a nonlinear function "f" using only the first two features
    # Note that I use only the first two features to calculate the labels
    y[i] = f(x[i,:2])
        
plt.figure(figsize = (5,5))
plt.scatter(x[:,0],x[:,1],c=y,s=4,cmap="winter")
plt.show

fn=filename("x_XGB_26") # file containing the data
print(fn)

np.savetxt(fn,x,fmt="%.3f")
fn=filename("y_XGB_26") # file containing the labels
np.savetxt(fn,y,fmt="%d")
print(fn)
```
