### Function for the random step
create the random increments of a time series as random walk 

$DX$ is the standard deviation, $bias$ is the constant average of the step

```python
# random seed for reproducibility
np.random.seed(12345)

# function for the random step, using lambda construction
# int() for cleaner look and for mimiking a detector with finite resolution
jump = lambda drift, stdev: int(np.random.normal(drift,stdev))

for i in range(10):
    print(jump(5,50))
```

### Function for the added pattern
to add to part of a time series, over $z$ bins, with amplitude $a$

```python
def pattern(i,z,a):
    return int(a*np.sin((np.pi*i)/z))
```

```python

## random seed for reproducibility
np.random.seed(12345)

# pattern parameters: Z=nr of steps, A=amplitude
Z=12
A=500

# number of data samples
N=10000
# size of each sample of the timeseries
L=60
# step parameters: introduce small positive bias 
DX = 50 # standard deviation of the random step
bias = 5  # bias to make the random walk drift in the positive direction

y = [0] * N # initialize class labels with zeros
x = [[0] * L for i in range(N)] # initialize timeseries with zeros

for i in range(N):
    if i>0:
        x[i][0] = x[i-1][-1] + jump(bias,DX) # first step of the new sample starts from the last point of the previous one
        # This means that time series (samples) are correlated by the first step
    
    for j in range(1,L):
        x[i][j] = x[i][j-1] + jump(bias,DX) # random walk with bias
    
    # Here I set class 0 (all time series with identified by an index that is a multiple of 3)
    y[i] = i%3 # assign class labels (0,1,2)
    ##y[i] = random.randint(0,2) # random assingment
    if y[i]>0:
        j0 = np.random.randint(0,L-1-Z) # random starting point for the pattern
        ###print(i,j0,j1)
        sign = 3-2*y[i] # sign of the pattern: + for class 1, - for class 2
        for j in range(Z):
            x[i][j0+j] += sign*pattern(j,Z,A) # add pattern to the timeseries

# The last line changes the random walk values, giving a distinguishable pattern that can be recognized. 
# It's like injecting a signal inside some random noise. If I have the right tools, I can distinguish between the pure noise and 
# the signal

for i in range(min(3,N)):
    print(x[i],y[i])
    

Show_data(x,L,"original data")

# Blue curve has y = 0 -> pure random walk
# Orange curve has y = 1 -> sign = +1 -> positive pattern (signal) injected
# Green curve has y = 2 -> sign = -1 -> negative pattern (signal) injected
```
